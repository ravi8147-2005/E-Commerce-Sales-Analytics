# Sales Analysis

### Business Objective

Analyze overall sales performance across products, customers, regions and time to identify business opportunities.

```python
total_sales=df['Sales'].sum()
print("Total Sales : ${:,.2f}".format(total_sales))
```

```python
total_profit=df['Profit'].sum()
print("Total Profit : ${:,.2f}".format(total_profit))
```

```python
total_orders=df['Order ID'].nunique()
print("Total Order : ", total_orders)
```

```python
total_customers=df['Customer ID'].nunique()
print("Total Customers :",total_customers)
```

```python
average_order_value = total_sales / total_orders
print("Average Order Value : ${:,.2f}".format(average_order_value))
```

```python
print("="*40)
print(" BUSINESS KPIs ")
print("="*40)

print(f"Total Sales : ${total_sales:,.2f}")

print(f"Total Profit : ${total_profit:,.2f}")

print(f"Total Orders : {total_orders}")

print(f"Total Customers : {total_customers}")

print(f"Average Order Value : ${average_order_value:,.2f}")
```

## Business Insights

### Key Findings

- The company generated total sales of **$2.30 Million**.

- Overall profit was **$286.40 Thousand**.

- Customers placed **5,009 unique orders**.
- The business served **793 unique customers**.
- The Average Order Value (AOV) was approximately **$458.62**, indicating the average revenue generated per customer order.

### Business Recommendations

- Increase Average Order Value through product bundles and cross-selling.
- Focus on retaining high-value customers to maximize revenue.
- Monitor profit margin alongside sales to ensure sustainable business growth.

# **Category Sales Analysis**

### **Sales by Category**
Which category generates the highest sales?

```python
category_sales=df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales
```

Category Sales Percentage

```python
category_sales_percentage = (category_sales / total_sales) * 100
category_sales_percentage.round(2)
```

Category Profit



```python
category_profit=df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
category_profit
```

Profit Margin

```python
#Created New DateFrame
category_analysis=pd.DataFrame({"Sales":category_sales,"Profit": category_profit})
```

```python
#created New column
category_analysis['Profit Margin (%)'] = (
    category_analysis['Profit']
    / category_analysis['Sales']
) * 100
#Display
category_analysis.round(2)
```

**Visualization**

```python
#Bar chart
plt.figure(figsize=(8,5))

category_sales.plot(
    kind='bar',
    edgecolor='black'
)

plt.title("Sales by Category", fontsize=14, fontweight='bold')
plt.xlabel("Category")
plt.ylabel("Sales ($)")
plt.xticks(rotation=0)

# Add value labels
for index, value in enumerate(category_sales):
    plt.text(index, value, f'{value:,.0f}', ha='center', va='bottom', fontsize=9)

plt.show()
```

## Business Insights

### Findings

- Technology generated the highest sales ($836,154), contributing 36.4% of total company revenue.

- Technology also achieved the highest profit ($145,455) with a profit margin of 17.40%.
- Office Supplies contributed 31.3% of sales and maintained a healthy profit margin of 17.04%.
- Furniture generated strong sales (32.3% of total revenue) but had a very low profit margin of only 2.49%.

### Recommendations

- Continue investing in Technology products because they generate both high sales and high profitability.
- Review Furniture pricing, discounts, and product costs to improve profitability.
- Maintain the current strategy for Office Supplies while identifying opportunities to increase sales.

- Most transactions were completed with either 0% or 20% discount.

- Avoid offering discounts above 20% unless they are part of a strategic campaign.
- Review products that require discounts of 30% or more.
- Optimize pricing instead of relying heavily on discounts.
- Analyze high-discount regions (such as Texas) to determine whether discounts are contributing to financial losses.

# Time Analysis

## Business Objective

Analyze yearly, monthly, and quarterly sales and profit trends to identify business growth patterns, seasonal demand, and peak-performing periods.

```python
#Converted the date columns
df['Order Date']=pd.to_datetime(df['Order Date'], format='mixed')
df['Ship Date']=pd.to_datetime(df['Ship Date'], format='mixed')
```

```python
#New columns
df['Year']=df['Order Date'].dt.year
df['Month']=df['Order Date'].dt.month_name()
df['Month Number']=df['Order Date'].dt.month
df['Quarter']=df['Order Date'].dt.quarter
```

```python
#Sales by Year
year_sales=df.groupby('Year')['Sales'].sum().sort_index()

year_sales
```

```python
#Profit by Year
year_profit=df.groupby('Year')['Profit'].sum().sort_index()

year_profit
```

```python
#Year Analysis Table
year_analysis=pd.DataFrame({
    "Sales": year_sales,
    "Profit":year_profit
})
year_analysis=year_analysis.round(2)
year_analysis
```

```python
#Sales by Month
month_sales=df.groupby(['Month Number', 'Month'])['Sales'].sum().reset_index().sort_values('Month Number')
month_sales
```

```python
#Profit by Month
month_profit=df.groupby(['Month Number','Month'])['Profit'].sum().reset_index().sort_values('Month Number')
month_profit
```

```python
#Quarter Analysis
quarter_sales=df.groupby('Quarter')[['Sales','Profit']].sum()
quarter_sales
```

# Visualization

```python
#Year Sales Chart
plt.figure(figsize=(8,5))
plt.plot(
    year_sales.index,
    year_sales.values,
    marker='o',
    linewidth=2
)
plt.title('Sales By Year',fontsize=14,fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Sales ($)')
plt.grid(True)
plt.show()
```

```python
#Monthly Sales Chart
plt.figure(figsize=(15,8))
plt.plot(
    month_sales['Month'],
    month_sales['Sales'],
    marker='o',
    linewidth=2
)
plt.title('Monthly Sales Trends',fontsize=14,fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid(True)
plt.show()
```

## Business Insights

### Findings

- Sales and profit showed strong year-over-year growth from 2015 to 2017.
- Although sales dipped slightly in 2015, profit increased, indicating better cost management or pricing.
- November generated the highest monthly sales and profit, followed by December.
- February recorded the lowest sales, indicating seasonal demand fluctuations.
- Quarter 4 (October–December) generated the highest sales and profit.

### Recommendations

- Increase inventory before the fourth quarter to meet higher customer demand.
- Launch marketing campaigns in October to capitalize on the strong holiday season.
- Analyze the reasons for lower sales during February and develop promotional strategies to improve performance.
- Continue optimizing pricing and operational efficiency to sustain profit growth.

# Shipping Analysis

```python
#Create Shipping Days
df['Shipping Days']=(df['Ship Date'] - df['Order Date']).dt.days

```

```python
#Shipping Mode Analysis
shipping_analysis = (
    df.groupby('Ship Mode')[['Sales', 'Profit']]
      .sum()
      .round(2)
)
shipping_analysis
```

```python
#Average Shipping Time
shipping_days = df.groupby('Ship Mode')['Shipping Days'].mean().round(2)
shipping_days
```

```python
#Combined Everything
shipping_report =shipping_analysis.copy()
shipping_report['Average Shipping Days']= shipping_days
shipping_report
```

# Visualazation

```python
plt.figure(figsize=(8,5))

shipping_report['Sales'].plot(
    kind='bar',
    edgecolor='black'
)

plt.title("Sales by Shipping Mode", fontsize=14, fontweight='bold')
plt.xlabel("Shipping Mode")
plt.ylabel("Sales ($)")
plt.xticks(rotation=0)

plt.show()
```

## Business Insights

### Findings

- Standard Class generated the highest sales ($1.36M)

*   and the highest profit ($164K).

- Same Day shipping generated the lowest sales and profit.

- Most customers preferred Standard Class shipping.

- Average shipping times matched the expected service levels.

### Recommendations

- Continue promoting Standard Class shipping because it balances customer demand and profitability.
- Investigate whether Same Day shipping can be made more attractive through premium services or targeted promotions.
- Monitor shipping efficiency to maintain customer satisfaction.
